```python
import pandas as pd
from pathlib import Path
ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / 'src').is_dir())

df = pd.read_csv(ROOT / "results" / "experiments_1_3.csv")
cmp = df.pivot(index=["id", "category"], columns="experiment", values="mrr")
cmp[cmp.nunique(axis=1) > 1]          # pitanja gde se eksperimenti razlikuju
```

```python
pd.set_option("display.max_colwidth", None)
diff_ids = cmp[cmp.nunique(axis=1) > 1].index.get_level_values("id")
df[df["id"].isin(diff_ids)][["id", "experiment", "mrr", "top5"]].sort_values(["id", "experiment"])
```

```python
import pandas as pd
from pathlib import Path
ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / 'src').is_dir())

df = pd.read_csv(ROOT / "results" / "experiments_1_3.csv")
d = df[(df["experiment"] == "3b_section_path_filter") & (df["category"] != "abstain")].copy()

# pozicija prvog tačnog izvora = 1 / MRR (MRR 0 znači: nije pronađen u top 10)
d["rank"] = d["mrr"].apply(lambda m: round(1 / m) if m > 0 else None)

print(d["rank"].value_counts(dropna=False).sort_index())
d[d["rank"] != 1][["id", "category", "rank", "top5"]]
```
